```python
lista = ['ID', 'Tipo', 'Título', 'Diretor', 'Elenco', 'País', 'Data de adição', 'Ano de lançamento', 'Classificação', 'duração', 'Listado em', 'Descrição']

Index = ['show_id', 'type', 'title', 'director', 'cast', 'country', 'date_added', 'release_year', 'rating', 'duration', 'listed_in', 'description']

rename = {}
for old, new in zip(Index, lista):
    rename[old]=new

rename
```

```python
df = df_netflix.rename(columns=rename)
df.head(3)
```

```python
df_netflix = df_netflix.rename(columns = {
    'show_id': 'ID',
    'type': 'Tipo',
    'title': 'Título',
    'director':'Diretor', 
    'cast':'Elenco', 
    'country':'País', 
    'date_added':'Data de adição',
    'release_year':'Ano de lançamento', 
    'rating':'Classificação', 
    'duration':'duração', 
    'listed_in':'Listado em', 
    'description':'Descrição'
})
df_netflix.head(3)
```

```python
df_netflix = df_netflix.drop(columns = ['Descrição'])
df_netflix.head(3)
```
